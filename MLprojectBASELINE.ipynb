```python
def extract_frames(video_path, output_dir, fps=1):
    os.makedirs(output_dir, exist_ok=True)
    cap = cv2.VideoCapture(video_path)

    video_fps = cap.get(cv2.CAP_PROP_FPS)
    frame_interval = int(video_fps // fps) if video_fps > fps else 1
    count = 0
    saved = 0

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        if count % frame_interval == 0:
            cv2.imwrite(os.path.join(output_dir, f"frame_{saved}.jpg"), frame)
            saved += 1
        count += 1
    cap.release()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, roc_curve, roc_auc_score, precision_recall_curve, auc
from sklearn.model_selection import StratifiedKFold
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def plot_confusion_matrix(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm)
    disp.plot(cmap='Blues')

def plot_roc_curve(y_true, y_probs):
    fpr, tpr, _ = roc_curve(y_true, y_probs)
    auc_score = roc_auc_score(y_true, y_probs)

    plt.plot(fpr, tpr, label=f"AUC = {auc_score:.2f}")
    plt.plot([0, 1], [0, 1], '--', color='gray')
    plt.xlabel("Predictions")
    plt.ylabel("Actual")
    plt.title("ROC Curve")
    plt.legend()
    plt.grid()
    plt.show()

def plot_precision_recall_curve(y_true, y_probs):
    precision, recall, _ = precision_recall_curve(y_true, y_probs)
    pr_auc = auc(recall, precision)

    plt.plot(recall, precision, label=f"PR AUC = {pr_auc:.2f}")
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("Precision-Recall Curve")
    plt.legend()
    plt.grid()
    plt.show()
```

```python
def cross_validate_model(model, X, y, n_splits=5):
    skf = StratifiedKFold(n_splits=n_splits)
    scores = []

    for train_index, test_index in skf.split(X, y):
        X_train, X_test = X[train_index], X[test_index]
        y_train, y_test = y[train_index], y[test_index]

        model.fit(X_train, y_train)
        y_probs = model.predict_proba(X_test)[:, 1]
        score = roc_auc_score(y_test, y_probs)
        scores.append(score)

    print(f"Mean ROC AUC over {n_splits} folds: {np.mean(scores):.4f}")
```

```python
# === Dataset ===
class FaceDataset(Dataset):
    """
    PyTorch Dataset for loading preprocessed face images and corresponding labels.

    Args:
        image_paths: List of file paths to face images
        labels: Corresponding list of labels (0=real, 1=fake)
        transform: torchvision transforms to apply to each image
    """
    def __init__(self, image_paths, labels, transform=None):
        self.image_paths = image_paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image = Image.open(self.image_paths[idx]).convert('RGB')
        if self.transform:
            image = self.transform(image)
        label = self.labels[idx]
        return image, label

# === Simple CNN ===
class SimpleCNN(nn.Module):
    """
    Simple Convolutional Neural Network for binary classification of face images.

    Layers:
        - Conv2d + ReLU + MaxPool
        - Conv2d + ReLU + MaxPool
        - Flatten + Linear + ReLU
        - Final Linear for binary output
    """
    def __init__(self):
        super(SimpleCNN, self).__init__()
        self.net = nn.Sequential(
            nn.Conv2d(3, 16, 3, 1, 1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, 1, 1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(32 * 28 * 28, 64), nn.ReLU(),
            nn.Linear(64, 1)
        )
    def forward(self, x):
        return self.net(x).squeeze(1)

# === Load Data Paths ===
def load_image_paths_and_labels(base_dir):
    """
    Load image file paths and generate binary labels based on directory names.

    Args:
        base_dir: Root directory containing 'real' and 'fake' subdirectories

    Returns:
        image_paths: List of file paths to images
        labels: List of corresponding binary labels (0=real, 1=fake)
    """
    image_paths, labels = [], []
    for subdir in os.listdir(base_dir):

        sub_path = os.path.join(base_dir, subdir)
        if os.path.isdir(sub_path):
            label = 0 if "real" in subdir else 1
            for img_file in glob.glob(os.path.join(sub_path, "**", "*.jpg"), recursive=True):
                image_paths.append(img_file)
                labels.append(label)
    return image_paths, labels

#James - the data structure I used to save the data was a bit different from yours so I rewrote this function
def load_image_paths_and_labels_new(base_dir):
    # Load the CSV (assumes columns: file, label)

    ####MAKE SURE THIS LINE OF CODE IS WHERE celeb_DF_results.csv is!!!!
    df = pd.read_csv("/content/drive/MyDrive/ML/celeb_df_results.csv")

    image_paths = []
    labels = []

    for _, row in df.iterrows():
        relative_path = row["file"]
        label = row["label"]

        #####MAKE SURE THIS LINE IS WHERE DIRECTORY WHERE IMAGES ARE STORED!!!!!!
        full_path = os.path.join("/content/drive/MyDrive/ML/frames_output2", relative_path)
        image_paths.append(full_path)
        labels.append(label)

    return image_paths, labels
```

```python
def train(config):
    """
    Train a binary classification model using preprocessed face images.

    Args:
        config: Dictionary loaded from a YAML config file containing hyperparameters

    Performs:
        - Loads face image data and splits into train/val sets
        - Initializes model and optimizer
        - Trains model for N epochs
        - Evaluates using accuracy and AUC
        - Logs metrics and model to MLflow
        - Saves training results to CSV

    Returns:
        log_df: Pandas DataFrame containing training metrics per epoch
    """
    # === Load from config ===
    base_dir = config["data"]["base_dir"]
    input_size = tuple(config["experiment"]["input_size"])
    batch_size = config["experiment"]["batch_size"]
    learning_rate = config["experiment"]["learning_rate"]
    epochs = config["experiment"]["epochs"]
    test_split = config["data"]["test_split"]
    log_file = config["output"]["log_file"]

    # === Load Data ===
    #my personal thing
    #image_paths, labels = load_image_paths_and_labels(base_dir)

    #base_dir = celeb_df_results
    #content/celeb_df_results.csv has image_name and label for all files
    #files are stored in /kaggle/working/frames_output
    #image_paths, labels = ???
    image_paths, labels = load_image_paths_and_labels_new(base_dir)
    train_paths, val_paths, train_labels, val_labels = train_test_split(
        image_paths, labels, test_size=test_split, stratify=labels
    )

    transform = transforms.Compose([
        transforms.Resize(input_size),
        transforms.ToTensor(),
    ])

    train_dataset = FaceDataset(train_paths, train_labels, transform)
    val_dataset = FaceDataset(val_paths, val_labels, transform)

    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = SimpleCNN().to(device)

    criterion = nn.BCEWithLogitsLoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)

    results_log = []

    with mlflow.start_run(run_name=config["experiment"]["name"]):
        for epoch in range(epochs):
            model.train()
            total_loss = 0

            for x_batch, y_batch in train_loader:
                x_batch = x_batch.to(device)
                y_batch = torch.tensor(y_batch, dtype=torch.float32).to(device)

                logits = model(x_batch)
                loss = criterion(logits, y_batch)
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
                total_loss += loss.item()

            model.eval()
            all_preds, all_labels = [], []

            with torch.no_grad():
                for x_batch, y_batch in val_loader:
                    x_batch = x_batch.to(device)
                    y_batch = torch.tensor(y_batch, dtype=torch.float32).to(device)
                    logits = model(x_batch)
                    probs = torch.sigmoid(logits).cpu().numpy()
                    all_preds.extend(probs)
                    all_labels.extend(y_batch.cpu().numpy())

            preds_bin = [1 if p > 0.5 else 0 for p in all_preds]
            acc = accuracy_score(all_labels, preds_bin)
            auc = roc_auc_score(all_labels, all_preds)

            # Log metrics to MLflow
            mlflow.log_metric("train_loss", total_loss, step=epoch)
            mlflow.log_metric("val_acc", acc, step=epoch)
            mlflow.log_metric("val_auc", auc, step=epoch)

            # Print + save metrics
            print(f"Epoch {epoch+1}: Loss={total_loss:.3f}, Val Acc={acc:.3f}, Val AUC={auc:.3f}")
            results_log.append({"epoch": epoch+1, "train_loss": total_loss, "val_acc": acc, "val_auc": auc})

            # Evaluation plots
            print("Confusion Matrix:")
            plot_confusion_matrix(all_labels, preds_bin)
            print("ROC Curve:")
            plot_roc_curve(all_labels, all_preds)
            print("Precision-Recall Curve:")
            plot_precision_recall_curve(all_labels, all_preds)

        # Log model + save results
        mlflow.pytorch.log_model(model, "model")
        log_df = pd.DataFrame(results_log)
        log_df.to_csv(log_file, index=False)

    return log_df
```

```python
# === Transfer Learning Models ===
import torchvision.models as models
import torch.nn as nn
import torch

class TransferLearningModel(nn.Module):
    """
    Transfer learning model using a pre-trained CNN backbone with a custom classification head.

    Args:
        model_name: Name of the pre-trained model to use (efficientnet_b0, resnet18, mobilenet_v2)
        freeze_backbone: Whether to freeze the backbone weights
        dropout_rate: Dropout rate for the classification head
    """
    def __init__(self, model_name='efficientnet_b0', freeze_backbone=True, dropout_rate=0.5):
        super(TransferLearningModel, self).__init__()
        self.model_name = model_name

        # Load pre-trained backbone
        if model_name == 'efficientnet_b0':
            self.backbone = models.efficientnet_b0(weights='DEFAULT')
            feature_dim = self.backbone.classifier[1].in_features
            self.backbone.classifier = nn.Identity()
        elif model_name == 'resnet18':
            self.backbone = models.resnet18(weights='DEFAULT')
            feature_dim = self.backbone.fc.in_features
            self.backbone.fc = nn.Identity()
        elif model_name == 'mobilenet_v2':
            self.backbone = models.mobilenet_v2(weights='DEFAULT')
            feature_dim = self.backbone.classifier[1].in_features
            self.backbone.classifier = nn.Identity()
        else:
            raise ValueError(f"Unsupported model: {model_name}")

        # Freeze backbone weights if requested
        if freeze_backbone:
            for param in self.backbone.parameters():
                param.requires_grad = False

        # Custom classification head
        self.classifier = nn.Sequential(
            nn.Dropout(dropout_rate),
            nn.Linear(feature_dim, 256),
            nn.ReLU(),
            nn.Dropout(dropout_rate),
            nn.Linear(256, 64),
            nn.ReLU(),
            nn.Linear(64, 1)
        )

    def forward(self, x):
        features = self.backbone(x)
        return self.classifier(features).squeeze(1)

    def unfreeze_layers(self, num_layers=3):
        """Unfreeze the last few layers of the backbone for fine-tuning"""
        if self.model_name == 'efficientnet_b0':
            modules = list(self.backbone.features.children())
            for module in modules[-num_layers:]:
                for param in module.parameters():
                    param.requires_grad = True
        elif self.model_name == 'resnet18':
            layers = [self.backbone.layer4, self.backbone.layer3, self.backbone.layer2]
            for i, layer in enumerate(layers):
                if i < num_layers:
                    for param in layer.parameters():
                        param.requires_grad = True
        elif self.model_name == 'mobilenet_v2':
            modules = list(self.backbone.features.children())
            for module in modules[-num_layers:]:
                for param in module.parameters():
                    param.requires_grad = True

# === Enhanced Face Dataset with Augmentation ===
class AugmentedFaceDataset(Dataset):
    """
    Enhanced dataset for face images with data augmentation.

    Args:
        image_paths: List of file paths to face images
        labels: Corresponding list of labels (0=real, 1=fake)
        transform: Base transforms to apply (normalization, ToTensor)
        augment: Whether to apply data augmentation
        phase: 'train' or 'val' to determine which transforms to apply
    """
    def __init__(self, image_paths, labels, transform=None, augment=True, phase='train'):
        self.image_paths = image_paths
        self.labels = labels
        self.transform = transform
        self.augment = augment
        self.phase = phase

        # Normalization values for ImageNet pre-trained models
        self.normalize = transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225]
        )

        # Training augmentations specifically designed for face images
        if augment and phase == 'train':
            self.train_transform = transforms.Compose([
                transforms.RandomHorizontalFlip(),  # Flip images (faces are symmetric)
                transforms.RandomRotation(10),      # Slight rotation
                transforms.ColorJitter(            # Color variations
                    brightness=0.2,
                    contrast=0.2,
                    saturation=0.2,
                    hue=0.1
                ),
                transforms.RandomAffine(         # Slight perspective changes
                    degrees=0,
                    translate=(0.1, 0.1),
                    scale=(0.9, 1.1)
                ),
            ])

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image = Image.open(self.image_paths[idx]).convert('RGB')

        # Apply base resizing transform
        if self.transform:
            image = self.transform(image)

        # Apply augmentations for training
        if self.augment and self.phase == 'train':
            image = self.train_transform(image)

        # Apply normalization (always)
        image = self.normalize(image)

        label = self.labels[idx]
        return image, label

# === Training Utils ===
class EarlyStopping:
    """
    Early stopping to prevent overfitting.

    Args:
        patience: Number of epochs to wait after last improvement
        delta: Minimum change to be considered as improvement
        path: Path to save the checkpoint
    """
    def __init__(self, patience=7, delta=0, path='checkpoint.pt'):
        self.patience = patience
        self.delta = delta
        self.path = path
        self.counter = 0
        self.best_score = None
        self.early_stop = False
        self.val_loss_min = float('inf')

    def __call__(self, val_loss, model):
        score = -val_loss

        if self.best_score is None:
            self.best_score = score
            self.save_checkpoint(val_loss, model)
        elif score < self.best_score + self.delta:
            self.counter += 1
            print(f'EarlyStopping counter: {self.counter} out of {self.patience}')
            if self.counter >= self.patience:
                self.early_stop = True
        else:
            self.best_score = score
            self.save_checkpoint(val_loss, model)
            self.counter = 0

    def save_checkpoint(self, val_loss, model):
        """Save model when validation loss decreases."""
        print(f'Validation loss decreased ({self.val_loss_min:.6f} --> {val_loss:.6f}). Saving model...')
        torch.save(model.state_dict(), self.path)
        self.val_loss_min = val_loss
```

```python
def train_transfer_model(config, model_name):
    """
    Train a transfer learning model using preprocessed face images.

    Args:
        config: Dictionary loaded from a YAML config file containing hyperparameters

    Returns:
        log_df: Pandas DataFrame containing training metrics per epoch
        best_model: The best model from training
    """
    # === Load from config ===
    base_dir = config["data"]["base_dir"]
    input_size = tuple(config["experiment"]["input_size"])
    batch_size = config["experiment"]["batch_size"]
    learning_rate = config["experiment"]["learning_rate"]
    epochs = config["experiment"]["epochs"]
    test_split = config["data"]["test_split"]
    log_file = config["output"]["log_file"]

    # New transfer learning parameters
    #model_name = config["transfer_learning"]["model_name"]
    freeze_backbone = config["transfer_learning"]["freeze_backbone"]
    dropout_rate = config["transfer_learning"]["dropout_rate"]
    patience = config["transfer_learning"]["patience"]
    unfreeze_after_epoch = config["transfer_learning"]["unfreeze_after_epoch"]

    # === Load Data ===
    image_paths, labels = load_image_paths_and_labels_new(base_dir)
    train_paths, val_paths, train_labels, val_labels = train_test_split(
        image_paths, labels, test_size=test_split, stratify=labels
    )

    # Base transform (resize to match pre-trained model requirements)
    transform = transforms.Compose([
        transforms.Resize(input_size),
        transforms.ToTensor(),
    ])

    # Create datasets with augmentation
    train_dataset = AugmentedFaceDataset(
        train_paths, train_labels, transform=transform, augment=True, phase='train'
    )
    val_dataset = AugmentedFaceDataset(
        val_paths, val_labels, transform=transform, augment=False, phase='val'
    )

    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)

    # Set up model
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = TransferLearningModel(
        model_name=model_name,
        freeze_backbone=freeze_backbone,
        dropout_rate=dropout_rate
    ).to(device)

    # Loss function
    criterion = nn.BCEWithLogitsLoss()

    # Optimizer with weight decay
    optimizer = optim.AdamW(
        model.parameters(),
        lr=learning_rate,
        weight_decay=1e-4
    )

    # Learning rate scheduler
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=3,
        verbose=True
    )

    # Early stopping
    early_stopping = EarlyStopping(
        patience=patience,
        path=f'best_{model_name}_model.pt'
    )

    results_log = []

    with mlflow.start_run(run_name=f"{config['experiment']['name']}_{model_name}"):
        # Log hyperparameters
        mlflow.log_params({
            "model_name": model_name,
            "freeze_backbone": freeze_backbone,
            "dropout_rate": dropout_rate,
            "learning_rate": learning_rate,
            "batch_size": batch_size,
            "input_size": input_size,
        })

        for epoch in range(epochs):
            # Unfreeze backbone layers for fine-tuning after specified epoch
            if epoch == unfreeze_after_epoch and freeze_backbone:
                print("Unfreezing last few layers of backbone for fine-tuning...")
                model.unfreeze_layers(num_layers=3)
                # Reset optimizer with unfrozen parameters (lower learning rate)
                optimizer = optim.AdamW(
                    [
                        {'params': [p for n, p in model.backbone.named_parameters() if p.requires_grad], 'lr': learning_rate/10},
                        {'params': model.classifier.parameters()}
                    ],
                    lr=learning_rate,
                    weight_decay=1e-4
                )
                scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3)

            # Training phase
            model.train()
            total_loss = 0
            train_preds, train_labels_list = [], []

            for x_batch, y_batch in train_loader:
                x_batch = x_batch.to(device)
                y_batch = torch.tensor(y_batch, dtype=torch.float32).to(device)

                logits = model(x_batch)
                loss = criterion(logits, y_batch)

                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

                total_loss += loss.item()

                # Track predictions for training metrics
                probs = torch.sigmoid(logits).cpu().detach().numpy()
                train_preds.extend(probs)
                train_labels_list.extend(y_batch.cpu().numpy())

            # Validation phase
            model.eval()
            val_loss = 0
            val_preds, val_labels_list = [], []

            with torch.no_grad():
                for x_batch, y_batch in val_loader:
                    x_batch = x_batch.to(device)
                    y_batch = torch.tensor(y_batch, dtype=torch.float32).to(device)

                    logits = model(x_batch)
                    loss = criterion(logits, y_batch)
                    val_loss += loss.item()

                    probs = torch.sigmoid(logits).cpu().numpy()
                    val_preds.extend(probs)
                    val_labels_list.extend(y_batch.cpu().numpy())

            # Calculate metrics
            train_preds_bin = [1 if p > 0.5 else 0 for p in train_preds]
            val_preds_bin = [1 if p > 0.5 else 0 for p in val_preds]

            train_acc = accuracy_score(train_labels_list, train_preds_bin)
            val_acc = accuracy_score(val_labels_list, val_preds_bin)
            train_auc = roc_auc_score(train_labels_list, train_preds)
            val_auc = roc_auc_score(val_labels_list, val_preds)

            # Update learning rate scheduler
            scheduler.step(val_loss)

            # Log metrics to MLflow
            mlflow.log_metric("train_loss", total_loss, step=epoch)
            mlflow.log_metric("val_loss", val_loss, step=epoch)
            mlflow.log_metric("train_acc", train_acc, step=epoch)
            mlflow.log_metric("val_acc", val_acc, step=epoch)
            mlflow.log_metric("train_auc", train_auc, step=epoch)
            mlflow.log_metric("val_auc", val_auc, step=epoch)

            # Print + save metrics
            print(f"Epoch {epoch+1}/{epochs}:")
            print(f"  Train Loss: {total_loss:.3f}, Train Acc: {train_acc:.3f}, Train AUC: {train_auc:.3f}")
            print(f"  Val Loss: {val_loss:.3f}, Val Acc: {val_acc:.3f}, Val AUC: {val_auc:.3f}")

            results_log.append({
                "epoch": epoch+1,
                "model": model_name,
                "train_loss": total_loss,
                "val_loss": val_loss,
                "train_acc": train_acc,
                "val_acc": val_acc,
                "train_auc": train_auc,
                "val_auc": val_auc
            })

            # Evaluation plots (every 5 epochs)
            if (epoch + 1) % 5 == 0:
                print("Confusion Matrix:")
                plot_confusion_matrix(val_labels_list, val_preds_bin)
                print("ROC Curve:")
                plot_roc_curve(val_labels_list, val_preds)
                print("Precision-Recall Curve:")
                plot_precision_recall_curve(val_labels_list, val_preds)

            # Early stopping
            early_stopping(val_loss, model)
            if early_stopping.early_stop:
                print("Early stopping triggered")
                break

        # Load best model
        model.load_state_dict(torch.load(f'best_{model_name}_model.pt'))

        # Log model
        mlflow.pytorch.log_model(model, f"model_{model_name}")

        # Save results
        log_df = pd.DataFrame(results_log)
        log_df.to_csv(f"{model_name}_{log_file}", index=False)

    return log_df, model

# Function to compare multiple architectures
def compare_architectures(config, models_to_compare=['efficientnet_b0', 'resnet18', 'mobilenet_v2']):
    """
    Compare different pre-trained model architectures.

    Args:
        config: Dictionary loaded from a YAML config file
        models_to_compare: List of model names to compare

    Returns:
        comparison_df: DataFrame with results from all models
    """
    results = []

    for model_name in models_to_compare:
        print(f"\n=== Training {model_name} ===\n")

        # Update config with current model
        #config["transfer_learning"]["model_name"] = model_name

        # Train model
        model_log_df, _ = train_transfer_model(config, model_name)

        # Store results
        results.append(model_log_df)

    # Combine results
    comparison_df = pd.concat(results)

    # Save comparison
    comparison_df.to_csv("model_comparison.csv", index=False)

    # Visualize comparisons
    best_metrics = comparison_df.groupby('model').agg({
        'val_acc': 'max',
        'val_auc': 'max'
    }).reset_index()

    print("\n=== Model Comparison Results ===")
    print(best_metrics)

    return comparison_df
```

```python
def download_dlib_models(download_dir='.'):
    """
    Download dlib models required for face detection and alignment.

    Args:
        download_dir: Directory where models will be saved

    Returns:
        Dict with paths to downloaded models
    """
    model_paths = {}

    # Make sure the directory exists
    os.makedirs(download_dir, exist_ok=True)

    # Face landmarks model
    landmarks_model_url = "https://github.com/davisking/dlib-models/raw/master/shape_predictor_68_face_landmarks.dat.bz2"
    landmarks_model_path = os.path.join(download_dir, "shape_predictor_68_face_landmarks.dat")
    landmarks_model_compressed = os.path.join(download_dir, "shape_predictor_68_face_landmarks.dat.bz2")

    # CNN face detector model (optional)
    cnn_model_url = "https://github.com/davisking/dlib-models/raw/master/mmod_human_face_detector.dat.bz2"
    cnn_model_path = os.path.join(download_dir, "mmod_human_face_detector.dat")
    cnn_model_compressed = os.path.join(download_dir, "mmod_human_face_detector.dat.bz2")

    # Download and extract landmarks model if needed
    if not os.path.exists(landmarks_model_path):
        print(f"Downloading facial landmarks model to {landmarks_model_path}...")
        try:
            # Download the compressed file
            urllib.request.urlretrieve(landmarks_model_url, landmarks_model_compressed)

            # Extract the file
            with open(landmarks_model_path, 'wb') as new_file, bz2.BZ2File(landmarks_model_compressed, 'rb') as file:
                shutil.copyfileobj(file, new_file)

            # Remove the compressed file
            os.remove(landmarks_model_compressed)
            print("Facial landmarks model downloaded successfully!")
            model_paths['landmarks'] = landmarks_model_path
        except Exception as e:
            print(f"Error downloading landmarks model: {e}")
            return None
    else:
        print(f"Facial landmarks model already exists at {landmarks_model_path}")
        model_paths['landmarks'] = landmarks_model_path

    # Ask if CNN model should be downloaded
    download_cnn = input("Do you want to download the CNN face detector model? (slower but more accurate) [y/N]: ")
    if download_cnn.lower() == 'y':
        if not os.path.exists(cnn_model_path):
            print(f"Downloading CNN face detector model to {cnn_model_path}...")
            try:
                # Download the compressed file
                urllib.request.urlretrieve(cnn_model_url, cnn_model_compressed)

                # Extract the file
                with open(cnn_model_path, 'wb') as new_file, bz2.BZ2File(cnn_model_compressed, 'rb') as file:
                    shutil.copyfileobj(file, new_file)

                # Remove the compressed file
                os.remove(cnn_model_compressed)
                print("CNN face detector model downloaded successfully!")
                model_paths['cnn'] = cnn_model_path
            except Exception as e:
                print(f"Error downloading CNN model: {e}")
        else:
            print(f"CNN face detector model already exists at {cnn_model_path}")
            model_paths['cnn'] = cnn_model_path

    return model_paths

# Download OpenCV DNN models if needed
def download_opencv_dnn_models(download_dir='.'):
    """
    Download OpenCV DNN model files for face detection.

    Args:
        download_dir: Directory where models will be saved

    Returns:
        Dict with paths to downloaded models
    """
    model_paths = {}

    # Make sure the directory exists
    os.makedirs(download_dir, exist_ok=True)

    # Model file URLs
    caffemodel_url = "https://github.com/opencv/opencv_3rdparty/raw/dnn_samples_face_detector_20170830/res10_300x300_ssd_iter_140000.caffemodel"
    prototxt_url = "https://github.com/opencv/opencv/raw/master/samples/dnn/face_detector/deploy.prototxt"

    # File paths
    caffemodel_path = os.path.join(download_dir, "res10_300x300_ssd_iter_140000.caffemodel")
    prototxt_path = os.path.join(download_dir, "deploy.prototxt")

    # Download model files if needed
    if not os.path.exists(caffemodel_path):
        print(f"Downloading OpenCV DNN caffemodel to {caffemodel_path}...")
        try:
            urllib.request.urlretrieve(caffemodel_url, caffemodel_path)
            print("Caffemodel downloaded successfully!")
            model_paths['caffemodel'] = caffemodel_path
        except Exception as e:
            print(f"Error downloading caffemodel: {e}")
            return None
    else:
        print(f"Caffemodel already exists at {caffemodel_path}")
        model_paths['caffemodel'] = caffemodel_path

    if not os.path.exists(prototxt_path):
        print(f"Downloading OpenCV DNN prototxt to {prototxt_path}...")
        try:
            urllib.request.urlretrieve(prototxt_url, prototxt_path)
            print("Prototxt downloaded successfully!")
            model_paths['prototxt'] = prototxt_path
        except Exception as e:
            print(f"Error downloading prototxt: {e}")
            return None
    else:
        print(f"Prototxt already exists at {prototxt_path}")
        model_paths['prototxt'] = prototxt_path

    return model_paths
```

```python
class FaceProcessor:
    def __init__(self,
                 detector_type='dlib_hog',  # Options: 'dlib_hog', 'dlib_cnn', 'opencv_haar', 'opencv_dnn'
                 predictor_path=None,
                 alignment_size=(112, 112),
                 quality_threshold=0.7,
                 blur_threshold=50,
                 batch_size=8):
        """
        Initialize face processor with detection, alignment, and tracking capabilities

        Args:
            detector_type: The type of detector to use
            predictor_path: Path to the facial landmark predictor model
            alignment_size: output size for aligned face
            quality_threshold: threshold for face quality score
            blur_threshold: threshold for blur detection (higher means less blur allowed)
            batch_size: batch size for processing
        """
        self.detector_type = detector_type
        self.alignment_size = alignment_size
        self.quality_threshold = quality_threshold
        self.blur_threshold = blur_threshold
        self.batch_size = batch_size

        # Initialize face detectors
        if detector_type == 'dlib_hog':
            # Dlib's HOG-based face detector
            self.detector = dlib.get_frontal_face_detector()
        elif detector_type == 'dlib_cnn':
            # Dlib's CNN-based face detector
            if not os.path.exists('mmod_human_face_detector.dat'):
                raise FileNotFoundError("CNN model file not found. Please download it using download_dlib_models() function.")
            self.detector = dlib.cnn_face_detection_model_v1('mmod_human_face_detector.dat')
        elif detector_type == 'opencv_haar':
            # OpenCV's Haar Cascade face detector
            self.detector = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
        elif detector_type == 'opencv_dnn':
            # OpenCV's DNN-based face detector
            if not os.path.exists('deploy.prototxt') or not os.path.exists('res10_300x300_ssd_iter_140000.caffemodel'):
                raise FileNotFoundError("DNN model files not found. Please download them using download_opencv_dnn_models() function.")
            self.detector = cv2.dnn.readNetFromCaffe(
                'deploy.prototxt',
                'res10_300x300_ssd_iter_140000.caffemodel'
            )
        else:
            raise ValueError(f"Unsupported detector: {detector_type}")

        # Initialize facial landmark predictor
        if predictor_path:
            try:
                self.predictor = dlib.shape_predictor(predictor_path)
            except RuntimeError:
                self.predictor = None
                print(f"Warning: Could not load the facial landmark predictor from {predictor_path}")
                print("Face alignment will not be available.")
        else:
            self.predictor = None
            print("No facial landmark predictor provided. Face alignment will not be available.")

        # Initialize tracking variables
        self.next_id = 0
        self.tracked_faces = {}
        self.face_history = defaultdict(list)

        # Reference landmarks for alignment (standard 5 points)
        self.reference_landmarks = np.array([
            [38.2946, 51.6963],  # left eye
            [73.5318, 51.5014],  # right eye
            [56.0252, 71.7366],  # nose tip
            [41.5493, 92.3655],  # left mouth corner
            [70.7299, 92.2041]   # right mouth corner
        ], dtype=np.float32)

        # Scale reference landmarks to target size
        self.reference_landmarks[:, 0] *= alignment_size[0] / 112
        self.reference_landmarks[:, 1] *= alignment_size[1] / 112

    def detect_faces(self, image):
        """
        Detect faces in an image using the configured detector

        Args:
            image: RGB image array

        Returns:
            list of face information dictionaries
        """
        faces = []
        h, w = image.shape[:2]

        try:
            if self.detector_type == 'dlib_hog':
                # Dlib's HOG detector
                gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
                dlib_faces = self.detector(gray, 1)

                for face in dlib_faces:
                    x1, y1 = face.left(), face.top()
                    x2, y2 = face.right(), face.bottom()

                    landmarks = None
                    if self.predictor:
                        shape = self.predictor(gray, face)
                        landmarks = np.array([[shape.part(i).x, shape.part(i).y] for i in range(68)])

                    face_info = {
                        'box': [x1, y1, x2, y2],
                        'landmarks': landmarks,
                        'confidence': 1.0  # HOG detector doesn't provide confidence scores
                    }
                    faces.append(face_info)

            elif self.detector_type == 'dlib_cnn':
                # Dlib's CNN detector
                dlib_faces = self.detector(image)

                for face in dlib_faces:
                    x1, y1 = face.rect.left(), face.rect.top()
                    x2, y2 = face.rect.right(), face.rect.bottom()
                    confidence = face.confidence

                    if confidence < 0:  # Skip if confidence is too low
                        continue

                    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
                    landmarks = None
                    if self.predictor:
                        shape = self.predictor(gray, face.rect)
                        landmarks = np.array([[shape.part(i).x, shape.part(i).y] for i in range(68)])

                    face_info = {
                        'box': [x1, y1, x2, y2],
                        'landmarks': landmarks,
                        'confidence': confidence
                    }
                    faces.append(face_info)

            elif self.detector_type == 'opencv_haar':
                # OpenCV's Haar Cascade detector
                gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
                opencv_faces = self.detector.detectMultiScale(
                    gray,
                    scaleFactor=1.1,
                    minNeighbors=5,
                    minSize=(30, 30)
                )

                for (x, y, w, h) in opencv_faces:
                    x1, y1, x2, y2 = x, y, x + w, y + h

                    landmarks = None
                    if self.predictor:
                        rect = dlib.rectangle(x1, y1, x2, y2)
                        shape = self.predictor(gray, rect)
                        landmarks = np.array([[shape.part(i).x, shape.part(i).y] for i in range(68)])

                    face_info = {
                        'box': [x1, y1, x2, y2],
                        'landmarks': landmarks,
                        'confidence': 1.0  # Haar detector doesn't provide confidence scores
                    }
                    faces.append(face_info)

            elif self.detector_type == 'opencv_dnn':
                # OpenCV's DNN detector
                blob = cv2.dnn.blobFromImage(
                    cv2.resize(image, (300, 300)),
                    1.0,
                    (300, 300),
                    (104.0, 177.0, 123.0)
                )
                self.detector.setInput(blob)
                detections = self.detector.forward()

                for i in range(detections.shape[2]):
                    confidence = detections[0, 0, i, 2]

                    if confidence < 0.5:  # Skip low confidence detections
                        continue

                    box = detections[0, 0, i, 3:7] * np.array([w, h, w, h])
                    x1, y1, x2, y2 = box.astype(int)

                    # Ensure coordinates are within image bounds
                    x1, y1 = max(0, x1), max(0, y1)
                    x2, y2 = min(w, x2), min(h, y2)

                    if x2 <= x1 or y2 <= y1:
                        continue

                    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
                    landmarks = None
                    if self.predictor:
                        rect = dlib.rectangle(x1, y1, x2, y2)
                        shape = self.predictor(gray, rect)
                        landmarks = np.array([[shape.part(i).x, shape.part(i).y] for i in range(68)])

                    face_info = {
                        'box': [x1, y1, x2, y2],
                        'landmarks': landmarks,
                        'confidence': float(confidence)
                    }
                    faces.append(face_info)
        except Exception as e:
            print(f"Error in face detection: {e}")

        return faces

    def align_face(self, image, landmarks):
        """
        Align face according to landmark positions

        Args:
            image: input image
            landmarks: facial landmarks (68 points from dlib)

        Returns:
            aligned face image
        """
        if landmarks is None or self.predictor is None:
            # If no landmarks or predictor, just return a crop of the face
            return image

        # Extract the 5 specific landmarks we need for alignment
        # These indices are for the standard 68-point dlib face landmarks
        alignment_landmarks = np.array([
            landmarks[36],  # Left eye left corner (36 in dlib's 68 point system)
            landmarks[45],  # Right eye right corner (45 in dlib's 68 point system)
            landmarks[30],  # Nose tip (30 in dlib's 68 point system)
            landmarks[48],  # Left mouth corner (48 in dlib's 68 point system)
            landmarks[54]   # Right mouth corner (54 in dlib's 68 point system)
        ], dtype=np.float32)

        # Calculate transformation matrix
        transform_matrix = cv2.getAffineTransform(
            alignment_landmarks[:3],  # Use just eyes and nose for alignment
            self.reference_landmarks[:3]
        )

        # Apply transformation
        aligned_face = cv2.warpAffine(
            image,
            transform_matrix,
            self.alignment_size,
            borderValue=(0, 0, 0)
        )

        return aligned_face

    def check_face_quality(self, face_img, face_box):
        """
        Check face quality (blur, size, etc.)

        Args:
            face_img: face image
            face_box: bounding box [x1, y1, x2, y2]

        Returns:
            quality_score, is_good_quality
        """
        # Check minimum face size
        box_width = face_box[2] - face_box[0]
        box_height = face_box[3] - face_box[1]
        min_face_size = min(box_width, box_height)

        if min_face_size < 40:  # Too small
            return 0.0, False

        # Check blur
        gray = cv2.cvtColor(face_img, cv2.COLOR_RGB2GRAY)
        blur_score = cv2.Laplacian(gray, cv2.CV_64F).var()

        if blur_score < self.blur_threshold:  # Too blurry
            return 0.5, False

        # Check face proportions
        ratio = box_width / box_height
        if ratio < 0.5 or ratio > 2.0:  # Face has strange proportions
            return 0.6, False

        # Simple quality score based on blur and size
        quality_score = min(1.0, (blur_score / 200) * (min_face_size / 100))

        return quality_score, quality_score > self.quality_threshold

    def track_face(self, face_info, frame_idx):
        """
        Track face across frames using IoU

        Args:
            face_info: face information dict
            frame_idx: current frame index

        Returns:
            face_id: unique identifier for the tracked face
        """
        current_box = face_info['box']

        # If no tracked faces yet, assign new ID
        if len(self.tracked_faces) == 0:
            face_id = self.next_id
            self.next_id += 1
            self.tracked_faces[face_id] = {
                'box': current_box,
                'last_seen': frame_idx
            }
            return face_id

        # Calculate IoU with all tracked faces
        max_iou = 0
        matched_id = None

        for face_id, face_data in self.tracked_faces.items():
            # Skip faces not seen in the last 30 frames
            if frame_idx - face_data['last_seen'] > 30:
                continue

            # Calculate IoU
            prev_box = face_data['box']
            iou = self._calculate_iou(current_box, prev_box)

            if iou > max_iou and iou > 0.3:  # Threshold for matching
                max_iou = iou
                matched_id = face_id

        # If match found, update tracked face
        if matched_id is not None:
            self.tracked_faces[matched_id]['box'] = current_box
            self.tracked_faces[matched_id]['last_seen'] = frame_idx
            return matched_id

        # No match found, create new track
        face_id = self.next_id
        self.next_id += 1
        self.tracked_faces[face_id] = {
            'box': current_box,
            'last_seen': frame_idx
        }
        return face_id

    def _calculate_iou(self, box1, box2):
        """
        Calculate Intersection over Union for two bounding boxes

        Args:
            box1: [x1, y1, x2, y2]
            box2: [x1, y1, x2, y2]

        Returns:
            IoU score (0-1)
        """
        x1_1, y1_1, x2_1, y2_1 = box1
        x1_2, y1_2, x2_2, y2_2 = box2

        # Calculate area of boxes
        area1 = (x2_1 - x1_1) * (y2_1 - y1_1)
        area2 = (x2_2 - x1_2) * (y2_2 - y1_2)

        # Calculate intersection
        x1_i = max(x1_1, x1_2)
        y1_i = max(y1_1, y1_2)
        x2_i = min(x2_1, x2_2)
        y2_i = min(y2_1, y2_2)

        if x2_i < x1_i or y2_i < y1_i:
            return 0.0  # No intersection

        intersection = (x2_i - x1_i) * (y2_i - y1_i)

        # Calculate union
        union = area1 + area2 - intersection

        # Calculate IoU
        iou = intersection / union if union > 0 else 0
        return iou

    def draw_faces(self, image, faces, draw_landmarks=True):
        """
        Draw bounding boxes and landmarks on the image

        Args:
            image: Image to draw on
            faces: List of face info dictionaries
            draw_landmarks: Whether to draw facial landmarks

        Returns:
            image with drawn faces
        """
        image_copy = image.copy()

        for face in faces:
            box = face['box']
            x1, y1, x2, y2 = box

            # Draw bounding box
            cv2.rectangle(image_copy, (x1, y1), (x2, y2), (0, 255, 0), 2)

            # Draw ID if available
            if 'id' in face:
                cv2.putText(image_copy, f"ID: {face['id']}", (x1, y1 - 10),
                           cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

            # Draw landmarks if available
            if draw_landmarks and 'landmarks' is not None and 'landmarks' in face and face['landmarks'] is not None:
                landmarks = face['landmarks']
                for (x, y) in landmarks:
                    cv2.circle(image_copy, (int(x), int(y)), 1, (0, 0, 255), -1)

        return image_copy

    def process_frame(self, frame, frame_idx):
        """
        Process a single frame

        Args:
            frame: RGB image
            frame_idx: frame index

        Returns:
            List of processed face information dictionaries
        """
        # Clone frame to avoid modifying original
        processed_frame = frame.copy()

        # Detect faces
        faces = self.detect_faces(processed_frame)
        result_faces = []

        for face_info in faces:
            box = face_info['box']
            landmarks = face_info['landmarks']

            # Extract face ROI
            x1, y1, x2, y2 = box
            x1, y1, x2, y2 = max(0, x1), max(0, y1), min(frame.shape[1], x2), min(frame.shape[0], y2)
            face_roi = processed_frame[y1:y2, x1:x2]

            if face_roi.size == 0:
                continue

            # Align face if landmarks are available
            aligned_face = self.align_face(processed_frame, landmarks) if landmarks is not None else face_roi

            # Check face quality
            quality_score, is_good_quality = self.check_face_quality(aligned_face, box)

            if not is_good_quality:
                continue

            # Track face
            face_id = self.track_face(face_info, frame_idx)

            # Store result
            result_face = {
                'id': face_id,
                'box': box,
                'landmarks': landmarks,
                'aligned_face': aligned_face,
                'quality_score': quality_score,
                'frame_idx': frame_idx
            }

            result_faces.append(result_face)

            # Store face in history
            self.face_history[face_id].append({
                'frame_idx': frame_idx,
                'box': box,
                'quality_score': quality_score,
                'aligned_face': aligned_face,
            })

        return result_faces

    def process_video(self, video_path, output_dir=None, max_frames=None, display_progress=True):
        """
        Process a video file and extract faces

        Args:
            video_path: path to video file
            output_dir: directory to save extracted faces (if None, faces are not saved to disk)
            max_frames: maximum number of frames to process (None for all)
            display_progress: whether to display progress bar

        Returns:
            face_dataset: dictionary of face IDs and their extracted face images
        """
        # Create output directory if specified
        if output_dir:
            os.makedirs(output_dir, exist_ok=True)

        # Open video
        cap = cv2.VideoCapture(video_path)

        if not cap.isOpened():
            raise ValueError(f"Could not open video: {video_path}")

        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        fps = cap.get(cv2.CAP_PROP_FPS)

        # Limit frames if needed
        if max_frames is not None:
            total_frames = min(total_frames, max_frames)

        # Reset tracking for new video
        self.next_id = 0
        self.tracked_faces = {}
        self.face_history = defaultdict(list)

        # Process frames
        frame_index = 0
        all_faces = []

        # Use tqdm for progress bar if requested
        iterator = tqdm(total=total_frames, desc="Processing frames") if display_progress else range(total_frames)

        while cap.isOpened() and (max_frames is None or frame_index < max_frames):
            ret, frame = cap.read()

            if not ret:
                break

            # Process at reduced frame rate for efficiency
            if frame_index % 3 == 0:  # Process every 3rd frame
                # Convert BGR to RGB
                frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

                # Process frame
                faces = self.process_frame(frame_rgb, frame_index)
                all_faces.extend(faces)

            frame_index += 1
            if display_progress:
                iterator.update(1)

        cap.release()
        if display_progress:
            iterator.close()

        # Filter and save the best face images for each ID
        face_dataset = {}

        for face_id, face_list in self.face_history.items():
            # Sort by quality score
            face_list.sort(key=lambda x: x['quality_score'], reverse=True)

            # Take top N faces
            top_faces = face_list[:min(10, len(face_list))]

            # Save faces if output_dir is specified
            if output_dir:
                id_dir = os.path.join(output_dir, f"id_{face_id}")
                os.makedirs(id_dir, exist_ok=True)

                for i, face_data in enumerate(top_faces):
                    face_img = face_data['aligned_face']
                    frame_idx = face_data['frame_idx']

                    # Convert RGB to BGR for saving
                    face_img_bgr = cv2.cvtColor(face_img, cv2.COLOR_RGB2BGR)

                    output_path = os.path.join(id_dir, f"frame_{frame_idx}_quality_{face_data['quality_score']:.2f}.jpg")
                    cv2.imwrite(output_path, face_img_bgr)

            face_dataset[face_id] = top_faces

        print(f"Processed {len(face_dataset)} unique faces")
        return face_dataset

    def display_face_dataset(self, face_dataset, num_examples=3):
        """
        Display the extracted faces in the notebook

        Args:
            face_dataset: Dictionary of face IDs and their face data
            num_examples: Number of example faces to show per identity
        """
        for face_id, face_list in face_dataset.items():
            print(f"Face ID: {face_id} - {len(face_list)} images")

            # Show the top examples
            fig, axes = plt.subplots(1, min(num_examples, len(face_list)), figsize=(15, 5))
            if min(num_examples, len(face_list)) == 1:
                axes = [axes]  # Make it iterable when there's only one image

            for i, face_data in enumerate(face_list[:num_examples]):
                face_img = face_data['aligned_face']
                quality = face_data['quality_score']
                frame_idx = face_data['frame_idx']

                axes[i].imshow(face_img)
                axes[i].set_title(f"Frame: {frame_idx}\nQuality: {quality:.2f}")
                axes[i].axis('off')

            plt.tight_layout()
            plt.show()

    def process_batch(self, video_paths, output_dir, max_frames_per_video=None, workers=4):
        """
        Process multiple videos in parallel

        Args:
            video_paths: list of video paths
            output_dir: base directory for outputs
            max_frames_per_video: maximum frames to process per video
            workers: number of parallel workers

        Returns:
            all_face_datasets: list of face datasets for each video
        """
        os.makedirs(output_dir, exist_ok=True)
        all_face_datasets = []

        # Define processing function for a single video
        def process_single_video(video_path):
            video_name = os.path.basename(video_path).split('.')[0]
            video_output_dir = os.path.join(output_dir, video_name)

            # Create a new processor instance for thread safety
            processor = FaceProcessor(detector_type=self.detector_type)

            try:
                face_dataset = processor.process_video(
                    video_path,
                    video_output_dir,
                    max_frames=max_frames_per_video,
                    display_progress=False  # Disable progress bars for parallel processing
                )
                return face_dataset
            except Exception as e:
                print(f"Error processing video {video_path}: {e}")
                return {}

        # Process videos in parallel
        with concurrent.futures.ThreadPoolExecutor(max_workers=workers) as executor:
            future_to_video = {
                executor.submit(process_single_video, video_path): video_path
                for video_path in video_paths
            }

            with tqdm(total=len(video_paths), desc="Processing videos") as pbar:
                for future in concurrent.futures.as_completed(future_to_video):
                    video_path = future_to_video[future]
                    try:
                        face_dataset = future.result()
                        all_face_datasets.append(face_dataset)
                    except Exception as e:
                        print(f"Error processing video {video_path}: {e}")
                    pbar.update(1)

        return all_face_datasets
```
